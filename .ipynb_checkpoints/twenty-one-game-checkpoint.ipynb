### Twenty-One Rules and Basics
In this project, you will be using a simplified version of Twenty-One, a turn-based, card-drawing game in which the goal is simple: Get your hand as close to 21 points as possible without going over. You can view the Gymnasium documentation of the environment [here](https://gymnasium.farama.org/environments/toy_text/blackjack/).

#### Game Setup
There are two players:
1) The agent (your learning algorithm)
2) The dealer (controlled by the environment)

Card values:
* Number cards: face value (e.g., 7 = 7 points)
* Face cards (J, Q, K): 10 points
* Aces: either 1 or 11, whichever helps the total

#### Observation Space
The agent observation space is a 3-tuple containing:
(player_sum, dealer_card, usable_ace)
* `player_sum`: Total value of the agent's hand (e.g., 18)
* `dealer_card`: The dealer's visible card (1–10)
* `usable_ace`: True or False, whether the agent has an ace that can count as 11 without going over 21

#### Actions
The agent chooses from:
* 0 = Hit → Take another card
* 1 = Stick → Stop drawing and end your turn

#### Episode
1) Both players start with two cards.
2) The agent makes a series of decisions: hit or stick.
3) If the agent's total exceeds 21, the episode ends and the agent loses immediately.
4) If the agent sticks, the dealer draws until reaching 17 or more.
5) The player closest to 21 without going over wins.

#### Rewards
* +1: Win
* 0: Tie
* –1: Loss

And that's Twenty-One in a nutshell-- good luck and happy learning!

## Environment Setup and Library Imports

First, let's import all necessary libraries and create a wrapper for the Twenty-One environment.

```python
# Import required libraries
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict
import random
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('default')
sns.set_palette("husl")

class TwentyOneEnv(gym.Wrapper):
    """
    Wrapper for the Blackjack environment to rename it as Twenty-One
    and provide consistent interface
    """
    def __init__(self, render_mode='rgb_array'):
        env = gym.make('Blackjack-v1', render_mode=render_mode)
        super().__init__(env)
        self.env_name = "TwentyOne-v1"
    
    def __str__(self):
        return f"<{self.env_name}>"
    
    def __repr__(self):
        return f"TwentyOneEnv({self.env_name})"

# Create the Twenty-One environment
def make_twenty_one_env(render_mode='rgb_array'):
    """Factory function to create Twenty-One environment"""
    return TwentyOneEnv(render_mode=render_mode)

print("Libraries imported successfully!")
print("Twenty-One environment wrapper created!")
print("\nEnvironment Details:")
print("- Game: Twenty-One (card game)")
print("- Actions: 0=Stick, 1=Hit")
print("- State: (player_sum, dealer_card, usable_ace)")
print("- Rewards: +1 (win), 0 (draw), -1 (loss)")
```

### Task 3: Single Episode Visualization
Create a function to play and visualize a single episode:
- Function name: `play_single_episode(env, policy_fn)`, which accepts the environment instance and policy function as parameters
- Use while loop for episode progression until termination
- Print each state, action taken, next state, and reward received
- Return the episode data list and total reward for analysis

```python
def play_single_episode(env, policy_fn):
    state = env.reset()[0]
    done = False
    total_reward = 0
    episode_data = []
    print(f"Starting state : {state}")
    while not done:
        action = policy_fn(state)
        next_state, reward, terminated, truncated, info = env.step(action)
        action_taken = "Hit" if action == 1 else "Stick"
        episode_data.append((state, action, reward))
        print(f'State: {state} , Action: {action_taken}, Reward: {reward}')
        state = next_state
        total_reward += reward
        done = terminated or truncated
    return episode_data, total_reward
```

### Task 4: Policy Testing
Define and test two different policies to understand basic strategies in Twenty-One:
- Create `random_policy(state)` using `np.random.choice([0, 1])` for random action selection
- Create `conservative_policy(state)` implementing the hit-when-sum-less-than-17 strategy
- Test both policies by running single episodes using the `play_single_episode()` function

```python
def random_policy(state):
    return np.random.choice([0,1])

def conservative_policy(state):
    player_sum = state[0]
    if state[0] < 17:
        return 1
    return 0

print("Random Policy Result")
episode, reward = play_single_episode(env, random_policy)

print("conservative policy Result")
episode, reward = play_single_episode(env, conservative_policy)
```

### Task 5: Analyze Win Rate by Policy
Now, to analyze the win rate by running multiple episodes to understand game dynamics:
- Create `analyze_state_distribution(env, policy_fn, episodes=1000)` function
- Track all states visited during episodes and corresponding episode rewards
- Return `states_visited` list and `rewards_received` list for visualization
- Apply analysis to random policy with 1000 episodes to establish baseline performance
- Calculate and display average reward and win rate metrics

```python
def analyze_state_distribution(env, policy_fn, episodes=1000):
    state_visited = []
    reward_received = []
    done = False 

    for i in range(episodes):
        state = env.reset()[0]
        done = False 
        episode_reward = 0

        while not done:
            action = policy_fn(state)
            next_state, reward, terminated, truncated, _ = env.step(action)
            done = terminated or truncated
            episode_reward += reward
            state_visited.append(state)
            state = next_state
        reward_received.append(episode_reward)
    return state_visited, reward_received

print("Analyzing random policy performance over 1000 episodes...")
states, rewards = analyze_state_distribution(env, random_policy, 1000)
print(f"Random policy - Average reward: {np.mean(rewards):.3f}, Win rate: {np.mean(np.array(rewards) > 0):.3f}")
print(f"Total states visited: {len(states)}")
```

## Task Group 2: Q-Learning and SARSA Implementation
### Task 7: State Mapping Functions
To prepare to implement tabular learning methods, you will need to use a state mapping to convert between Twenty-One's tuple state representation and integer indices required for tabular Q-learning.  Review the following function.
- Generate `state_to_index` dictionary for state→integer conversion
- Generate `index_to_state` dictionary for integer→state conversion
- Calculate and display total state space size (360 states)

```python
def create_state_mapping():
    state_to_index = {}
    index_to_state = {}
    index = 0
    
    for player_sum in range(4, 22):  # 4-21
        for dealer_card in range(1, 11):  # 1-10 (A-10)
            for usable_ace in [False, True]:
                state_to_index[(player_sum, dealer_card, usable_ace)] = index
                index_to_state[index] = (player_sum, dealer_card, usable_ace)
                index += 1
    
    return state_to_index, index_to_state

# Test the mapping
state_to_index, index_to_state = create_state_mapping()
print(f"Total number of states: {len(state_to_index)}")
print(f"Sample mapping: {(12, 5, True)} -> {state_to_index[(12, 5, True)]}")
print(f"Reverse mapping: {state_to_index[(12, 5, True)]} -> {index_to_state[state_to_index[(12, 5, True)]]}")
print(" State mapping functions created and tested!")
```

### Task 8: Complete Q-Learning Loop
Implement the Q-learning training loop:
- Use the parameters `episodes=50000, alpha=0.1, gamma=0.95, epsilon=0.1, epsilon_decay=0.9995, min_epsilon=0.01`
- Initialize the Q-table as zeroes with dimensions `(360, 2)`
- Implement epsilon-greedy action selection within the training loop
- Apply the Q-learning update rule using the maximum Q-value of the next state
- Include an epsilon decay mechanism with exponential decay and a minimum threshold
- Track metrics: rewards, wins, Q-value changes, epsilon values
- Provide progress reporting every 10,000 episodes with win rate and average reward
- Make sure to save the Q-table and metrics with unique variables to use later!

```python
def q_agent(env, episodes=50000, alpha=0.1, gamma=0.95, epsilon=0.1, epsilon_decay=0.9995, min_epsilon=0.01):
    state_to_index, index_to_state = create_state_mapping()
    
    q_table = np.zeros((len(state_to_index), env.action_space.n))
    
    wins = 0
    rewards = []
    q_deltas = []
    epsilons = []  

    for episode in range(episodes):
        state = env.reset()[0]
        done = False
        q_old_table = q_table.copy()  
        total_reward = 0

        while not done:
            state_index = state_to_index[state]
            
            if random.random() < epsilon:
                action = env.action_space.sample()
            else:
                action = np.argmax(q_table[state_index])
                
            next_state, reward, terminated, truncated, _ = env.step(action)
            done = terminated or truncated
            total_reward += reward

            if not done:
                next_state_idx = state_to_index[next_state]
                best_next_q = np.max(q_table[next_state_idx])
            else:
                best_next_q = 0
                
            q_table[state_index, action] += alpha * (reward + gamma * best_next_q - q_table[state_index, action])
            state = next_state
            
        rewards.append(total_reward)
        if total_reward > 0:
            wins += 1
            
        q_delta = np.mean(np.abs(q_table - q_old_table))
        q_deltas.append(q_delta)
        epsilons.append(epsilon)
        
        epsilon = max(min_epsilon, epsilon * epsilon_decay)
        
        if (episode + 1) % 10000 == 0:
            win_rate = wins / 10000
            avg_reward = np.mean(rewards[-10000:])
            print(f"Episode {episode+1}: Win Rate = {win_rate:.3f}, Avg Reward = {avg_reward:.3f}, Epsilon = {epsilon:.3f}")
            wins = 0 
            
    return q_table, state_to_index, rewards, q_deltas, epsilons
```

### Task 9: Complete SARSA Learning
Implement the complete SARSA learning with the same parameters, but using the SARSA updates:
- Use parameters `episodes=50000, alpha=0.1, gamma=0.95, epsilon=0.1, epsilon_decay=0.9995, min_epsilon=0.01)`
- Initialize Q-table and metrics tracking identical to Q-learning for fair comparison
- Implement distinctive SARSA action selection: choose both current and next actions using epsilon-greedy
- Apply SARSA update rule using Q-value of actual next action
- Include identical metrics tracking and progress reporting as Q-learning
- Make sure to save the Q-table and metrics with unique variables to use later!


```python
def sarsa_agent(env, episodes=50000, alpha=0.1, gamma=0.95, epsilon=0.1, epsilon_decay=0.9995, min_epsilon=0.01):
    state_to_index, index_to_state = create_state_mapping()
    q_table = np.zeros((len(state_to_index), env.action_space.n))
    
    rewards = []
    wins = 0
    q_deltas = []
    epsilons = []
    
    print(f"SARSA initialized with {len(state_to_index)} states and {env.action_space.n} actions")
    print(f"Training parameters: episodes={episodes}, alpha={alpha}, gamma={gamma}, epsilon={epsilon}")
    
    for episode in range(episodes):
        state, _ = env.reset()
        done = False
        total_reward = 0
        old_q_table = q_table.copy()
        
        state_idx = state_to_index[state]
        if random.random() < epsilon:
            action = env.action_space.sample()
        else:
            action = np.argmax(q_table[state_idx])
        
        while not done:
            next_state, reward, terminated, truncated, _ = env.step(action)
            done = terminated or truncated
            total_reward += reward
            
            if not done:
                next_state_idx = state_to_index[next_state]
                if random.random() < epsilon:
                    next_action = env.action_space.sample()
                else:
                    next_action = np.argmax(q_table[next_state_idx])
                
                q_table[state_idx, action] += alpha * (reward + gamma * q_table[next_state_idx, next_action] - q_table[state_idx, action])
                
                state = next_state
                state_idx = next_state_idx
                action = next_action
            else:
                q_table[state_idx, action] += alpha * (reward - q_table[state_idx, action])
        
        rewards.append(total_reward)
        if total_reward > 0:
            wins += 1
        
        q_delta = np.mean(np.abs(q_table - old_q_table))
        q_deltas.append(q_delta)
        epsilons.append(epsilon)
        
        epsilon = max(min_epsilon, epsilon * epsilon_decay)
        
        if (episode + 1) % 10000 == 0:
            win_rate = wins / 10000
            avg_reward = np.mean(rewards[-10000:])
            print(f"Episode {episode+1}: Win Rate = {win_rate:.3f}, Avg Reward = {avg_reward:.3f}, Epsilon = {epsilon:.3f}")
            wins = 0
    
    return q_table, state_to_index, rewards, q_deltas, epsilons
```

## Task Group 3: Policy Evaluation and Visualization

### Task 10: Policy Evaluation Function
Create a comprehensive policy evaluation function to test learned policies using pure greedy action selection:
- Function signature: `evaluate_policy(env, q_table, state_to_index, episodes=10000)`
- Use greedy action selection (no exploration) with `np.argmax(q_table[state_idx])` for true policy performance
- Track win rate (percentage of episodes with positive rewards) and average reward per episode
- Run evaluation over 10,000 episodes for statistical significance
- Display comprehensive results, including win rate percentage and performance analysis
- Return both metrics for comparative analysis between algorithms

```python
def evaluate_policy(env, q_table, state_to_index, episodes=10000):
    wins = 0
    total_rewards = 0
    
    for _ in range(episodes):
        state, _ = env.reset()
        done = False
        episode_reward = 0
        
        while not done:
            state_idx = state_to_index[state]
            action = np.argmax(q_table[state_idx])  # Always take best action
            state, reward, terminated, truncated, _ = env.step(action)
            done = terminated or truncated
            episode_reward += reward
        
        if episode_reward > 0:
            wins += 1
        total_rewards += episode_reward
    
    win_rate = wins / episodes
    avg_reward = total_rewards / episodes
    
    print(f"Policy Evaluation Results ({episodes} episodes):")
    print(f"Win Rate: {win_rate:.3f} ({win_rate*100:.1f}%)")
    print(f"Average Reward: {avg_reward:.3f}")
    
    return win_rate, avg_reward
```

### Task 11: Policy Heatmap Visualization
Create comprehensive policy heatmap visualization showing optimal actions across all game states using dual-panel layout:
- Function signature: `plot_policy_heatmap(q_table, state_to_index, title)`
- Create separate 18×10 matrices for usable ace and no usable ace scenarios
- Use seaborn heatmaps with proper axis labels: dealer cards (A,2-10) and player sums (4-21)
- Apply color coding with 'RdYlBu_r' colormap where 0=Stick (blue) and 1=Hit (red)
- Include annotations showing action values in each cell
- Add color bars with descriptive labels and professional formatting
- Use dual-panel layout to separate usable ace scenarios for strategic analysis

```python
def plot_policy_heatmap(q_table, state_to_index, title="Policy Heatmap"):
    u_grid = np.zeros((18, 10))
    n_grid = np.zeros((18, 10))

    for (p_sum, d_card, ace), idx in state_to_index.items():
        act = np.argmax(q_table[idx])
        r, c = p_sum - 4, d_card - 1
        if ace:
            u_grid[r, c] = act
        else:
            n_grid[r, c] = act

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
    x_labs = ["A"] + [str(i) for i in range(2, 11)]
    y_labs = list(range(4, 22))

    sns.heatmap(
        u_grid,
        annot=True,
        fmt="g",
        cmap="coolwarm",
        ax=ax1,
        xticklabels=x_labs,
        yticklabels=y_labs,
    )
    ax1.set_title(f"{title} - Ace")

    sns.heatmap(
        n_grid,
        annot=True,
        fmt="g",
        cmap="coolwarm",
        ax=ax2,
        xticklabels=x_labs,
        yticklabels=y_labs,
    )
    ax2.set_title(f"{title} - No Ace")

    plt.tight_layout()
    plt.show()
```

### Task 12: Training Metrics Visualization
Create comprehensive training metrics visualization to analyze convergence and learning behavior using multi-panel layout:
- Function signature: `plot_training_metrics(rewards, q_deltas, epsilons, title)`
- Create 2x2 subplot grid showing: moving average rewards, Q-value changes (log scale), epsilon decay, and win rate progression
- Use 1000-episode moving windows for smoothing noisy training curves
- Apply logarithmic scale for Q-value changes to visualize convergence effectively
- Include reference lines (zero reward, good performance threshold) for context
- Add professional formatting with grids, colors, and descriptive titles
- Enable comparative analysis between Q-learning and SARSA training behavior

```python
def plot_training_metrics(rewards, q_deltas, epsilons, title="Metrics"):
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))

    w = 1000
    avg = [np.mean(rewards[max(0, i - w) : i + 1]) for i in range(len(rewards))]
    ax1.plot(avg, "b")
    ax1.set_title("Avg Reward")

    ax2.plot(q_deltas, "g", alpha=0.6)
    ax2.set_yscale("log")
    ax2.set_title("Q Changes")

    ax3.plot(epsilons, "orange")
    ax3.set_title("Epsilon")

    rates = []
    for i in range(0, len(rewards), w):
        b = rewards[i : i + w]
        if b:
            rates.extend([sum(1 for r in b if r > 0) / len(b)] * len(b))

    ax4.plot(rates[: len(rewards)], "m")
    ax4.set_ylim(0, 1)
    ax4.set_title("Win Rate")

    plt.suptitle(title)
    plt.tight_layout()
    plt.show()
```

```python
from PIL import Image as PILImage
from IPython.display import HTML, display

# 1. Create the game environment
v_env = gym.make("Blackjack-v1", render_mode="rgb_array")
state, _ = v_env.reset()
done = False

# 2. Setup the notebook display hook handle
dh = display(display_id=True)

while not done:
    # Look up agent's best move
    idx = state_mapping[state]
    action = np.argmax(q_table_ql[idx])

    # Convert the raw array frame into a raw memory image format
    frame = v_env.render()
    img = PILImage.fromarray(frame)
    buffer = BytesIO()
    img.save(buffer, format="PNG")

    # Encode the image data into HTML string parts directly
    b64_str = base64.b64encode(buffer.getvalue()).decode("utf-8")
    html_frame = f'<img src="data:image/png;base64,{b64_str}" width="500"/>'

    # Instantly swap the image framework in place inside the same cell
    dh.update(HTML(html_frame))
    time.sleep(0.5)

    # Take the step
    state, reward, term, trunc, _ = v_env.step(action)
    done = term or trunc

# Render final end-of-game scene
frame = v_env.render()
img = PILImage.fromarray(frame)
buffer = BytesIO()
img.save(buffer, format="PNG")
b64_str = base64.b64encode(buffer.getvalue()).decode("utf-8")
dh.update(HTML(f'<img src="data:image/png;base64,{b64_str}" width="500"/>'))

print(f"Game Concluded! Agent Score Reward: {reward}")
v_env.close()
```
